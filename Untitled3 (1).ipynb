```python
Q.1 Text Analytics Engine
Write a Python program to analyse a paragraph entered by the user.
(a) Clean the text by converting it to a common case, removing punctuation and extra spaces, and
tokenize it using string methods.
(b) Create a word-frequency dictionary and display the top 5 most frequent words and the number of
unique words.
(c) Find the longest word, average word length and all palindromic words.
(d) Display the results as a properly formatted report.
(e) Handle empty input appropriately.
Restriction: Do not use re or collections
```

```python
def analyze_text(text):
    # Check for empty input
    if text.strip() == "":
        print("Error: Please enter a valid paragraph.")
        return

    # Convert text to lowercase
    text = text.lower()

    # Remove punctuation
    punctuation = ".,!?;:'\"()[]{}"
    for ch in punctuation:
        text = text.replace(ch, "")

    # Remove extra spaces and tokenize
    words = text.split()

    # Create word frequency dictionary
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

    # Top 5 frequent words
    top_words = sorted(frequency.items(),
                       key=lambda x: x[1], reverse=True)[:5]

    # Unique words
    unique_words = len(frequency)

    # Longest word
    longest_word = max(words, key=len)

    # Average word length
    average_length = sum(len(word) for word in words) / len(words)

    # Find palindrome words
    palindromes = []
    for word in frequency:
        if len(word) > 1 and word == word[::-1]:
            palindromes.append(word)

    # Display report
    print("\n----- TEXT ANALYSIS REPORT -----")
    print("Total words:", len(words))
    print("Unique words:", unique_words)
    print("Word frequency:", frequency)
    print("Top 5 frequent words:", top_words)
    print("Longest word:", longest_word)
    print("Average word length:", round(average_length, 2))
    print("Palindromic words:", palindromes)


def main():
    paragraph = input("Enter a paragraph: ")
    analyze_text(paragraph)


main()
```

```python
Q.2 Matrix Operations Using Nested Lists
Write a Python program to perform basic matrix operations using nested lists.
(a) Read two matrices from the user and validate their dimensions.
(b) Implement matrix addition and subtraction using user-defined functions.
(c) Implement matrix multiplication with appropriate dimension checking.
(d) Implement a function to calculate the transpose of a matrix.
(e) Display meaningful error messages when an operation is not possible.
Restriction: Do not use NumPy or any built-in matrix library.
```

```python
def read_matrix(rows, cols, name):
    print("Enter elements of", name)
    matrix = []

    for i in range(rows):
        row = []
        for j in range(cols):
            value = int(input("Enter element: "))
            row.append(value)
        matrix.append(row)

    return matrix


# Matrix Addition
def addition(A, B):
    if len(A) != len(B) or len(A[0]) != len(B[0]):
        print("Addition is not possible.")
        return

    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] + B[i][j])
        result.append(row)

    return result


# Matrix Subtraction
def subtraction(A, B):
    if len(A) != len(B) or len(A[0]) != len(B[0]):
        print("Subtraction is not possible.")
        return

    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(A[0])):
            row.append(A[i][j] - B[i][j])
        result.append(row)

    return result


# Matrix Multiplication
def multiplication(A, B):
    if len(A[0]) != len(B):
        print("Multiplication is not possible.")
        return

    result = []
    for i in range(len(A)):
        row = []
        for j in range(len(B[0])):
            total = 0
            for k in range(len(B)):
                total += A[i][k] * B[k][j]
            row.append(total)
        result.append(row)

    return result


# Matrix Transpose
def transpose(A):
    result = []
    for j in range(len(A[0])):
        row = []
        for i in range(len(A)):
            row.append(A[i][j])
        result.append(row)

    return result


def main():
    r1 = int(input("Enter rows of first matrix: "))
    c1 = int(input("Enter columns of first matrix: "))

    r2 = int(input("Enter rows of second matrix: "))
    c2 = int(input("Enter columns of second matrix: "))

    if r1 <= 0 or c1 <= 0 or r2 <= 0 or c2 <= 0:
        print("Invalid matrix dimensions.")
        return

    A = read_matrix(r1, c1, "Matrix A")
    B = read_matrix(r2, c2, "Matrix B")

    print("\nMatrix A:", A)
    print("Matrix B:", B)

    print("\nAddition:", addition(A, B))
    print("Subtraction:", subtraction(A, B))
    print("Multiplication:", multiplication(A, B))
    print("Transpose of A:", transpose(A))
    print("Transpose of B:", transpose(B))


main()
```

```python
Q.3 Student Record Management
Create a student record management program using nested dictionaries.
Each record should contain roll number, name, department and marks in three subjects.
(a) Add, update and delete student records while preventing duplicate roll numbers.
(b) Calculate total, percentage, grade and pass/fail status. A student fails if marks in any subject are
below 35.
(c) Display the student with the highest percentage and generate a rank list.
(d) Display the department-wise average percentage.
(e) Provide a simple menu and handle invalid choices.
```

```python
students = {}

# Add Student
def add_student():
    roll = input("Enter roll number: ")

    if roll in students:
        print("Roll number already exists!")
        return

    name = input("Enter student name: ")
    dept = input("Enter department: ")

    m1 = int(input("Enter marks in Subject 1: "))
    m2 = int(input("Enter marks in Subject 2: "))
    m3 = int(input("Enter marks in Subject 3: "))

    students[roll] = {
        "name": name,
        "department": dept,
        "marks": [m1, m2, m3]
    }

    print("Student added successfully!")


# Calculate Result
def calculate_result(roll):
    marks = students[roll]["marks"]

    total = sum(marks)
    percentage = total / 3

    if min(marks) < 35:
        grade = "F"
        status = "Fail"
    elif percentage >= 90:
        grade = "A"
        status = "Pass"
    elif percentage >= 75:
        grade = "B"
        status = "Pass"
    elif percentage >= 60:
        grade = "C"
        status = "Pass"
    else:
        grade = "D"
        status = "Pass"

    return total, percentage, grade, status


# Update Student
def update_student():
    roll = input("Enter roll number to update: ")

    if roll not in students:
        print("Student not found!")
        return

    name = input("Enter new name: ")
    dept = input("Enter new department: ")

    m1 = int(input("Enter new Subject 1 marks: "))
    m2 = int(input("Enter new Subject 2 marks: "))
    m3 = int(input("Enter new Subject 3 marks: "))

    students[roll] = {
        "name": name,
        "department": dept,
        "marks": [m1, m2, m3]
    }

    print("Student updated successfully!")


# Delete Student
def delete_student():
    roll = input("Enter roll number to delete: ")

    if roll in students:
        del students[roll]
        print("Student deleted successfully!")
    else:
        print("Student not found!")


# Display All Students
def display_students():
    if not students:
        print("No student records available.")
        return

    for roll in students:
        total, percentage, grade, status = calculate_result(roll)

        print("\nRoll Number:", roll)
        print("Name:", students[roll]["name"])
        print("Department:", students[roll]["department"])
        print("Total Marks:", total)
        print("Percentage:", round(percentage, 2))
        print("Grade:", grade)
        print("Status:", status)


# Display Topper and Rank List
def rank_list():
    if not students:
        print("No student records available.")
        return

    ranking = []

    for roll in students:
        total, percentage, grade, status = calculate_result(roll)
        ranking.append((roll, percentage))

    ranking.sort(key=lambda x: x[1], reverse=True)

    print("\n----- RANK LIST -----")

    for i, (roll, percentage) in enumerate(ranking, start=1):
        print(i, students[roll]["name"], round(percentage, 2), "%")

    topper = ranking[0][0]
    print("\nTopper:", students[topper]["name"])


# Department-wise Average
def department_average():
    departments = {}

    for roll in students:
        dept = students[roll]["department"]
        total, percentage, grade, status = calculate_result(roll)

        if dept not in departments:
            departments[dept] = []

        departments[dept].append(percentage)

    print("\n----- DEPARTMENT AVERAGE -----")

    for dept in departments:
        avg = sum(departments[dept]) / len(departments[dept])
        print(dept, ":", round(avg, 2), "%")


# Main Menu
def main():
    while True:
        print("\n----- STUDENT RECORD MANAGEMENT -----")
        print("1. Add Student")
        print("2. Update Student")
        print("3. Delete Student")
        print("4. Display Students")
        print("5. Rank List and Topper")
        print("6. Department-wise Average")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_student()
        elif choice == "2":
            update_student()
        elif choice == "3":
            delete_student()
        elif choice == "4":
            display_students()
        elif choice == "5":
            rank_list()
        elif choice == "6":
            department_average()
        elif choice == "7":
            print("Exiting program...")
            break
        else:
            print("Invalid choice! Please try again.")


main()
```

```python
Q.4 Recursion Laboratory
Write recursive functions for the following problems.
(a) Implement Tower of Hanoi and display all moves and the total number of moves.
(b) Implement recursive binary search on a sorted list and return the position of the searched element.
(c) Generate all unique permutations of a string.
(d) Demonstrate all functions using suitable test cases.
Restriction: The main solution for the above tasks must use recursion
```

```python
# Q4: Recursion Laboratory

# (a) Tower of Hanoi
def tower_of_hanoi(n, source, auxiliary, destination):
    if n == 1:
        print("Move disk 1 from", source, "to", destination)
        return 1

    moves = tower_of_hanoi(n - 1, source, destination, auxiliary)

    print("Move disk", n, "from", source, "to", destination)

    moves += 1
    moves += tower_of_hanoi(n - 1, auxiliary, source, destination)

    return moves


# (b) Recursive Binary Search
def binary_search(arr, low, high, key):
    if low > high:
        return -1

    mid = (low + high) // 2

    if arr[mid] == key:
        return mid
    elif key < arr[mid]:
        return binary_search(arr, low, mid - 1, key)
    else:
        return binary_search(arr, mid + 1, high, key)


# (c) Unique Permutations
def permutations(s):
    if len(s) == 0:
        return [""]

    result = []

    for i in range(len(s)):
        if s[i] in s[:i]:
            continue

        remaining = s[:i] + s[i + 1:]

        for p in permutations(remaining):
            result.append(s[i] + p)

    return result


# (d) Main function
def main():
    print("TOWER OF HANOI")
    n = 3
    total = tower_of_hanoi(n, 'A', 'B', 'C')
    print("Total moves:", total)

    print("\nBINARY SEARCH")
    arr = [10, 20, 30, 40, 50]
    key = 30
    position = binary_search(arr, 0, len(arr) - 1, key)
    print("List:", arr)
    print("Search element:", key)
    print("Position:", position)

    print("\nUNIQUE PERMUTATIONS")
    s = "AAB"
    print("String:", s)
    print("Permutations:", permutations(s))


main()
```

```python
Q5. Course Recommendation Using Sets
A university stores the courses taken by different users in the following form:
{user : {courses}}.
(a) Create records for at least 6 users and allow a user to add or remove a course.
(b) Calculate the Jaccard similarity between two users based on their course sets.
(c) For a selected user, recommend up to 3 courses taken by similar users but not taken by the
selected user.
(d) Find the most popular course among all users.
(e) Handle an invalid user name appropriately
```

```python
# Q5: Course Recommendation Using Sets

users = {
    "A": {"Python", "Java", "DBMS"},
    "B": {"Python", "C", "DBMS"},
    "C": {"Java", "Python", "AI"},
    "D": {"C", "DBMS", "AI"},
    "E": {"Python", "AI", "Java"},
    "F": {"C", "Java", "DBMS"}
}


def add_course(user, course):
    if user not in users:
        print("Invalid user!")
        return

    users[user].add(course)
    print("Course added successfully.")


def remove_course(user, course):
    if user not in users:
        print("Invalid user!")
        return

    if course in users[user]:
        users[user].remove(course)
        print("Course removed.")
    else:
        print("Course not found.")


def jaccard(user1, user2):
    if user1 not in users or user2 not in users:
        print("Invalid user!")
        return None

    s1 = users[user1]
    s2 = users[user2]

    union = s1 | s2
    intersection = s1 & s2

    if len(union) == 0:
        return 0

    return len(intersection) / len(union)


def recommend(user):
    if user not in users:
        print("Invalid user!")
        return

    scores = {}

    for other in users:
        if other != user:
            score = jaccard(user, other)

            for course in users[other] - users[user]:
                scores[course] = scores.get(course, 0) + score

    recommendations = sorted(
        scores, key=scores.get, reverse=True
    )[:3]

    print("Recommended courses:", recommendations)


def popular_course():
    frequency = {}

    for courses in users.values():
        for course in courses:
            frequency[course] = frequency.get(course, 0) + 1

    popular = max(frequency, key=frequency.get)

    print("Most popular course:", popular)
    print("Number of users:", frequency[popular])


def main():
    print("Users:", users)

    print("\nJaccard Similarity A and B:")
    print(round(jaccard("A", "B"), 2))

    recommend("A")

    popular_course()

    add_course("A", "AI")
    remove_course("A", "Java")

    # Invalid user test
    recommend("X")


main()
```

```python
Q6. Sales Data File Processing
A file sales.txt contains records in the format
ID,Region,Product,Price,Quantity
Write a program to process the file.
(a) Read the file safely and handle a missing-file error.
(b) Validate the records and ignore invalid rows such as negative quantities or non-numeric prices.
(c) Calculate the total revenue for each region.
(d) Find the region having the highest revenue.
(e) Write the valid records and a short summary to an output file
```

```python
# Q6: Sales Data File Processing

def process_sales():
    revenue = {}
    valid_records = []

    try:
        with open("sales.txt", "r") as file:
            lines = file.readlines()

    except FileNotFoundError:
        print("Error: sales.txt not found!")
        return

    for line in lines:
        try:
            data = line.strip().split(",")

            if len(data) != 5:
                continue

            ID, region, product, price, quantity = data

            price = float(price)
            quantity = int(quantity)

            if price < 0 or quantity < 0:
                continue

            amount = price * quantity

            revenue[region] = revenue.get(region, 0) + amount

            valid_records.append(line.strip())

        except ValueError:
            print("Ignoring invalid record:", line.strip())

    if not revenue:
        print("No valid sales records.")
        return

    highest_region = max(revenue, key=revenue.get)

    print("\nRegional Revenue:")
    for region, amount in revenue.items():
        print(region, ":", amount)

    print("Highest Revenue Region:", highest_region)

    with open("sales_output.txt", "w") as file:
        file.write("Valid Records:\n")

        for record in valid_records:
            file.write(record + "\n")

        file.write("\nRegional Revenue:\n")

        for region, amount in revenue.items():
            file.write(f"{region}: {amount}\n")

        file.write("Highest Revenue Region: " + highest_region)

    print("Output saved to sales_output.txt")


process_sales()
```

```python
Q7. Bank Transaction Processor
An input file transactions.txt contains transactions in the form
Account,Type,Amount
where the transaction type may be deposit or withdraw.
(a) Read and parse the transactions from the file.
(b) Update account balances for valid deposits and withdrawals.
(c) Reject withdrawals when the account does not have sufficient balance.
(d) Create and use at least one custom exception for an invalid transaction.
(e) Display final balances and the number of rejected transactions
```

```python
# Q7: Bank Transaction Processor

class InvalidTransaction(Exception):
    pass


def process_transactions():
    balances = {}
    rejected = 0

    try:
        with open("transactions.txt", "r") as file:
            transactions = file.readlines()

    except FileNotFoundError:
        print("transactions.txt not found!")
        return

    for line in transactions:
        try:
            account, typ, amount = line.strip().split(",")

            amount = float(amount)

            if amount <= 0:
                raise InvalidTransaction("Invalid amount")

            if typ == "deposit":
                balances[account] = balances.get(account, 0) + amount

            elif typ == "withdraw":
                if balances.get(account, 0) < amount:
                    raise InvalidTransaction("Insufficient balance")

                balances[account] -= amount

            else:
                raise InvalidTransaction("Invalid transaction type")

        except (ValueError, InvalidTransaction) as e:
            print("Rejected:", line.strip(), "-", e)
            rejected += 1

    print("\nFinal Balances:")
    for account, balance in balances.items():
        print(account, ":", balance)

    print("Rejected Transactions:", rejected)


def main():
    process_transactions()


main()
```

```python
Q8. Library Management System
Create the classes Book, Member and Library.
(a) Create the classes with suitable attributes and constructors.
(b) Implement book issue and return operations. A member can have a maximum of 3 books.
(c) Prevent a book from being issued if it is already issued.
(d) Calculate a fine of Rs. 2 per day for books returned after the due period.
(e) Display the list of issued books and members having fines
```

```python
# Q8: Library Management System

class Book:
    def __init__(self, book_id, title):
        self.book_id = book_id
        self.title = title
        self.issued = False


class Member:
    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name
        self.books = []
        self.fine = 0


class Library:
    def __init__(self):
        self.books = {}
        self.members = {}

    def add_book(self, book):
        self.books[book.book_id] = book

    def add_member(self, member):
        self.members[member.member_id] = member

    def issue_book(self, book_id, member_id):
        if book_id not in self.books:
            print("Book not found!")
            return

        if member_id not in self.members:
            print("Member not found!")
            return

        book = self.books[book_id]
        member = self.members[member_id]

        if book.issued:
            print("Book already issued!")
        elif len(member.books) >= 3:
            print("Maximum 3 books allowed!")
        else:
            book.issued = True
            member.books.append(book_id)
            print("Book issued successfully!")

    def return_book(self, book_id, member_id, late_days):
        if member_id not in self.members:
            print("Member not found!")
            return

        member = self.members[member_id]

        if book_id not in member.books:
            print("Book was not issued to this member!")
            return

        member.books.remove(book_id)
        self.books[book_id].issued = False

        fine = max(0, late_days) * 2
        member.fine += fine

        print("Book returned successfully!")
        print("Fine:", fine)

    def display_issued(self):
        print("\nIssued Books:")

        for book in self.books.values():
            if book.issued:
                print(book.book_id, book.title)

    def display_fines(self):
        print("\nMembers Having Fines:")

        for member in self.members.values():
            if member.fine > 0:
                print(member.name, ":", member.fine)


def main():
    lib = Library()

    lib.add_book(Book(1, "Python"))
    lib.add_book(Book(2, "Java"))
    lib.add_book(Book(3, "DBMS"))
    lib.add_book(Book(4, "C Programming"))

    lib.add_member(Member(101, "Rahul"))
    lib.add_member(Member(102, "Amit"))

    lib.issue_book(1, 101)
    lib.issue_book(2, 101)
    lib.issue_book(3, 101)

    lib.issue_book(4, 101)  # Maximum 3 books

    lib.return_book(1, 101, 5)

    lib.display_issued()
    lib.display_fines()


main()
```

```python
Q10. Secure Wallet System
Create a Wallet class demonstrating encapsulation.
(a) Create private attributes for balance and PIN.
(b) Implement deposit() and withdraw() with suitable validation.
(c) Implement PIN verification and change
pin().
(d) Lock the wallet after three consecutive incorrect PIN attempts.
(e) Display a simple transaction history or mini-statement.
Restriction: The PIN must not be directly exposed through a public attribute.
```

```python
# Q10: Secure Wallet System

class Wallet:
    def __init__(self, pin, balance=0):
        self.__balance = balance
        self.__pin = str(pin)
        self.__attempts = 0
        self.__locked = False
        self.__history = []

    def verify_pin(self, pin):
        if self.__locked:
            print("Wallet is locked!")
            return False

        if str(pin) == self.__pin:
            self.__attempts = 0
            return True

        self.__attempts += 1
        print("Incorrect PIN!")

        if self.__attempts >= 3:
            self.__locked = True
            print("Wallet locked after 3 incorrect attempts!")

        return False

    def deposit(self, amount):
        if self.__locked:
            print("Wallet is locked!")
            return

        if amount <= 0:
            print("Invalid amount!")
            return

        self.__balance += amount
        self.__history.append("Deposit: " + str(amount))

        print("Deposit successful!")

    def withdraw(self, amount, pin):
        if not self.verify_pin(pin):
            return

        if amount <= 0:
            print("Invalid amount!")
        elif amount > self.__balance:
            print("Insufficient balance!")
        else:
            self.__balance -= amount
            self.__history.append("Withdraw: " + str(amount))
            print("Withdrawal successful!")

    def change_pin(self, old_pin, new_pin):
        if not self.verify_pin(old_pin):
            return

        if len(str(new_pin)) != 4 or not str(new_pin).isdigit():
            print("PIN must contain 4 digits!")
            return

        self.__pin = str(new_pin)
        print("PIN changed successfully!")

    def show_balance(self, pin):
        if self.verify_pin(pin):
            print("Balance:", self.__balance)

    def mini_statement(self):
        if self.__locked:
            print("Wallet is locked!")
            return

        print("\nTransaction History:")

        for transaction in self.__history:
            print(transaction)

        print("Current Balance:", self.__balance)


def main():
    wallet = Wallet("1234", 5000)

    wallet.deposit(1000)
    wallet.withdraw(500, "1234")
    wallet.change_pin("1234", "5678")
    wallet.withdraw(200, "5678")
    wallet.show_balance("5678")
    wallet.mini_statement()

    # Test incorrect PIN
    wallet.withdraw(100, "0000")
    wallet.withdraw(100, "0000")
    wallet.withdraw(100, "0000")


main()
```

```python
class Polynomial:
    def __init__(self, coefficients):
        self.coeff = coefficients

    def __str__(self):
        terms = []
        for power, value in enumerate(self.coeff):
            if value != 0:
                terms.append(f"{value}x^{power}")
        return " + ".join(terms[::-1]) or "0"

    def __call__(self, x):
        return sum(c * (x ** i) for i, c in enumerate(self.coeff))

    def __add__(self, other):
        n = max(len(self.coeff), len(other.coeff))
        result = []
        for i in range(n):
            a = self.coeff[i] if i < len(self.coeff) else 0
            b = other.coeff[i] if i < len(other.coeff) else 0
            result.append(a + b)
        return Polynomial(result)

    def __sub__(self, other):
        n = max(len(self.coeff), len(other.coeff))
        result = []
        for i in range(n):
            a = self.coeff[i] if i < len(self.coeff) else 0
            b = other.coeff[i] if i < len(other.coeff) else 0
            result.append(a - b)
        return Polynomial(result)

    def __mul__(self, other):
        result = [0] * (len(self.coeff) + len(other.coeff) - 1)
        for i, a in enumerate(self.coeff):
            for j, b in enumerate(other.coeff):
                result[i + j] += a * b
        return Polynomial(result)

    def derivative(self):
        result = [i * self.coeff[i] for i in range(1, len(self.coeff))]
        return Polynomial(result or [0])


p1 = Polynomial([1, 2, 3])      # 3x² + 2x + 1
p2 = Polynomial([2, 1])         # x + 2
p3 = Polynomial([1, 0, 1])      # x² + 1

print("P1:", p1)
print("P2:", p2)
print("P3:", p3)
print("Addition:", p1 + p2)
print("Subtraction:", p1 - p2)
print("Multiplication:", p1 * p2)
print("P1(2):", p1(2))
print("Derivative:", p1.derivative())
```

```python
Q13. Inventory Manager
Create an inventory management system using OOP.
(a) Create a Product class containing product code, name, price and quantity with suitable validation.
(b) Create an Inventory class to store multiple products.
(c) Implement restock() and sell() operations. Selling more items than available stock must be
rejected.
(d) Calculate the total inventory value and display products whose quantity is below a specified limit.
(e) Save the current inventory details to a text file
```

```python
class Product:
    def __init__(self, code, name, price, quantity):
        if not code or not name:
            raise ValueError("Code and name required")
        if price < 0 or quantity < 0:
            raise ValueError("Invalid price or quantity")

        self.code = code
        self.name = name
        self.price = price
        self.quantity = quantity


class Inventory:
    def __init__(self):
        self.products = {}

    def add_product(self, product):
        if product.code in self.products:
            print("Product already exists")
        else:
            self.products[product.code] = product
            print("Product added")

    def restock(self, code, qty):
        if qty <= 0 or code not in self.products:
            print("Invalid restock")
            return
        self.products[code].quantity += qty

    def sell(self, code, qty):
        if code not in self.products or qty <= 0:
            print("Invalid sale")
            return
        product = self.products[code]
        if qty > product.quantity:
            print("Insufficient stock")
            return
        product.quantity -= qty
        print("Sale successful")

    def total_value(self):
        return sum(p.price * p.quantity
                   for p in self.products.values())

    def low_stock(self, limit=5):
        for p in self.products.values():
            if p.quantity < limit:
                print(p.code, p.name, p.quantity)

    def save(self, filename):
        with open(filename, "w") as f:
            for p in self.products.values():
                f.write(f"{p.code},{p.name},{p.price},{p.quantity}\n")


inv = Inventory()
inv.add_product(Product("P1", "Pen", 10, 20))
inv.add_product(Product("P2", "Book", 50, 4))

inv.restock("P1", 10)
inv.sell("P2", 2)

print("Total Inventory Value:", inv.total_value())
print("Low Stock Products:")
inv.low_stock()
inv.save("inventory.txt")
```

```python
class Doctor:
    def __init__(self, doctor_id, name):
        self.doctor_id = doctor_id
        self.name = name


class Patient:
    def __init__(self, patient_id, name):
        self.patient_id = patient_id
        self.name = name


class Appointment:
    appointments = []

    def __init__(self, doctor, patient, date, time):
        self.doctor = doctor
        self.patient = patient
        self.date = date
        self.time = time

    def book(self):
        for a in Appointment.appointments:
            if (a.doctor.doctor_id == self.doctor.doctor_id
                    and a.date == self.date and a.time == self.time):
                print("Slot already booked")
                return
        Appointment.appointments.append(self)
        print("Appointment booked")

    def cancel(self):
        if self in Appointment.appointments:
            Appointment.appointments.remove(self)
            print("Appointment cancelled")
        else:
            print("Appointment not found")

    @staticmethod
    def doctor_schedule(doctor):
        for a in Appointment.appointments:
            if a.doctor.doctor_id == doctor.doctor_id:
                print(a.date, a.time, a.patient.name)


d1 = Doctor("D1", "Dr. Sharma")
p1 = Patient("P1", "Amit")
p2 = Patient("P2", "Sneha")

a1 = Appointment(d1, p1, "2026-11-10", "10:00")
a2 = Appointment(d1, p2, "2026-11-10", "10:00")

a1.book()
a2.book()
Appointment.doctor_schedule(d1)
a1.cancel()
```

```python
Q15. Signal Processing with NumPy
Use NumPy to analyse a noisy signal.
(a) Generate 300 samples of a sine wave with random noise using a fixed random seed.
(b) Apply a moving-average filter using NumPy operations.
(c) Calculate the mean and standard deviation and identify values whose absolute z-score is greater
than 2.
(d) Replace detected outliers using interpolation or a suitable NumPy-based method.
(e) Plot the original and processed signals in the same figure and label them properly.
Restriction: Avoid explicit Python loops for the main NumPy operations
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

# Generate signal
t = np.linspace(0, 10, 300)
signal = np.sin(t)
noise = np.random.normal(0, 0.2, 300)
data = signal + noise

# Add artificial outliers
data[50] = 3
data[150] = -3

# Moving average
window = 5
moving_avg = np.convolve(
    data, np.ones(window) / window, mode="same"
)

# Z-score
mean = np.mean(data)
std = np.std(data)
z_score = (data - mean) / std

# Detect outliers
outliers = np.abs(z_score) > 2

# Interpolate outliers
processed = data.copy()
processed[outliers] = np.interp(
    np.where(outliers)[0],
    np.where(~outliers)[0],
    data[~outliers]
)

print("Mean:", mean)
print("Standard Deviation:", std)
print("Outliers:", np.where(outliers)[0])

# Object-oriented plotting
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(t, data, label="Original Signal")
ax.plot(t, moving_avg, label="Moving Average")
ax.plot(t, processed, label="Processed Signal")

ax.set_title("Signal Processing")
ax.set_xlabel("Time")
ax.set_ylabel("Amplitude")
ax.legend()
ax.grid()

plt.show()
```

```python
Q18. Plot Class Framework
Create a simple object-oriented plotting framework.
(a) Create a base class BasePlot containing data, title and axis-label information.
(b) Create two subclasses, LinePlot and BarPlot, each implementing its own render(ax) method.
(c) Create a Dashboard class that receives plot objects and displays them using subplots.
(d) Use polymorphism so that the Dashboard does not need to check the type of each plot object.
(e) Demonstrate the framework with at least one line plot and one bar plot and save the figure
```

```python
import matplotlib.pyplot as plt

class BasePlot:
    def render(self, ax):
        pass


class LinePlot(BasePlot):
    def __init__(self, x, y, title):
        self.x = x
        self.y = y
        self.title = title

    def render(self, ax):
        ax.plot(self.x, self.y, marker="o")
        ax.set_title(self.title)


class BarPlot(BasePlot):
    def __init__(self, labels, values, title):
        self.labels = labels
        self.values = values
        self.title = title

    def render(self, ax):
        ax.bar(self.labels, self.values)
        ax.set_title(self.title)


class Dashboard:
    def __init__(self):
        self.plots = []

    def add_plot(self, plot):
        self.plots.append(plot)

    def display(self):
        n = len(self.plots)
        if n == 0:
            print("No plots available")
            return

        fig, axes = plt.subplots(1, n, figsize=(5*n, 4))

        if n == 1:
            axes = [axes]

        for plot, ax in zip(self.plots, axes):
            plot.render(ax)

        fig.tight_layout()
        fig.savefig("oop_dashboard.png")
        plt.show()


dashboard = Dashboard()

dashboard.add_plot(
    LinePlot([1, 2, 3], [10, 20, 15], "Line Plot")
)

dashboard.add_plot(
    BarPlot(["A", "B", "C"], [30, 20, 40], "Bar Plot")
)

dashboard.display()
```

```python
Q20. Numerical Integration and Error Analysis
Create an object-oriented numerical integration program.
(a) Create a base class Integrator and subclasses Trapezoid and Midpoint.
(b) Implement the integration method for both subclasses.
(c) Evaluate a given function using n = 4,8,16,32 subintervals.
(d) Calculate the absolute error by comparing the numerical result with the exact integral.
(e) Plot error versus number of subintervals and write a short observation about how the error
changes
```

```python
import numpy as np
import matplotlib.pyplot as plt

class Integrator:
    def __init__(self, func, a, b):
        self.func = func
        self.a = a
        self.b = b

    def integrate(self, n):
        raise NotImplementedError


class Trapezoid(Integrator):
    def integrate(self, n):
        h = (self.b - self.a) / n
        x = np.linspace(self.a, self.b, n + 1)
        y = self.func(x)

        return h * (
            y[0] / 2 + np.sum(y[1:-1]) + y[-1] / 2
        )


class Midpoint(Integrator):
    def integrate(self, n):
        h = (self.b - self.a) / n
        x = self.a + (np.arange(n) + 0.5) * h

        return h * np.sum(self.func(x))


# Function: x^2 from 0 to 1
f = lambda x: x**2
exact = 1 / 3

methods = [
    Trapezoid(f, 0, 1),
    Midpoint(f, 0, 1)
]

n_values = [4, 8, 16, 32]

fig, ax = plt.subplots()

for method in methods:
    errors = []

    print("\n", method.__class__.__name__)

    for n in n_values:
        result = method.integrate(n)
        error = abs(exact - result)
        errors.append(error)

        print("n =", n, "Result =", result,
              "Error =", error)

    ax.plot(n_values, errors, marker="o",
            label=method.__class__.__name__)

ax.set_title("Numerical Integration Error")
ax.set_xlabel("Number of Subintervals (n)")
ax.set_ylabel("Absolute Error")
ax.legend()
ax.grid()

plt.show()
```
